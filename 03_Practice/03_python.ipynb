Write a function that determines whether a given object is of Python's str type.

Return:

True if the input is a string.
False otherwise.

Assumptions:

The input can be of any Python data type.
Use Python's built-in type-checking functionality.
Built-in functions are allowed.

```python
def is_string(value):
    return type(value) == str
    
result = is_string("hello")
print(result)

def is_string1(value):
    return isinstance(value, str)
    
result1 = is_string1(1)
print(result1)
```

```python
class MyString(str):
    pass

s = MyString("hello")

print(type(s) == str)
print(isinstance(s, str))
```

write a function that takes a value and returns the name of its data type as a string.

Return one of the following:

"int"
"float"
"str"
"bool"
"list"
"tuple"
"dict"
"set"
"NoneType"
"unknown" (for any other type)

Assumptions:

The input can be any Python object.
Built-in functions are allowed.

```python
def get_type_name(value):
    if type(value) == int:
        return "int"
    elif type(value) == float:
        return "float"
    elif type(value) == str:
        return "str"
    elif type(value) == bool:
        return "bool"
    elif type(value) == list:
        return "list"
    elif type(value) == tuple:
        return "tuple"
    elif type(value) == dict:
        return "dict"
    elif type(value) == set:
        return "set"
    elif value is None:
        return "NoneType" 
    else:
        return "unknown"

get_type_name(None)
    
```
