```python
def StockPrices(symbol):
    url = f'https://finance.yahoo.com/quote/{symbol}/'
    header = {'User-Agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/149.0.0.0 Safari/537.36'}
    r = requests.get(url,headers=header)
    soup = BeautifulSoup(r.text,'html.parser')
    price = soup.find("span", {"data-testid": "qsp-price"}).text
    p_change = soup.find("span", {"data-testid": "qsp-price-change"}).text
    per_change = soup.find("span", {"data-testid": "qsp-price-change-percent"}).text
    details = {'Symbol':symbol, 'price': price, 'Price Change': p_change, 'Percent Change' : per_change}
    return details
```

```python
StockPrices('AAL')
```

```python
# List of stock symbols
MyStocks = ['AAPL','MSFT','NFLX','ADBE','UBER']
StockData = []
```

```python
for symbol in MyStocks:
    StockData.append(StockPrices(symbol))
    print('Gettting price of Awesome Stock:', symbol)
StockData
```

```python

def FetchPrice():
    MyStocks = ['AAPL','MSFT','NFLX','ADBE','UBER']
    StockData = []
    for symbol in MyStocks:
        StockData.append(StockPrices(symbol))
        print('Gettting price of Awesome Stock:', symbol)
    return StockData
```

```python
# Automate

while True:
    print(FetchPrice())
    time.sleep(5)
    
```
